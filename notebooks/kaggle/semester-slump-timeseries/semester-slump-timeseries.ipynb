# 📉 The Semester 3–4 slump
Panel/time-series analysis of `semester_records.csv`: DSA, OS and DBMS arrive, grades drop, then senioritis hits.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns, warnings
warnings.filterwarnings('ignore'); sns.set_theme(style='whitegrid')
D = pd.read_csv('/kaggle/input/datasets/pranavsinghdeveloper/desi-cs-student-survival/students.csv')
S = pd.read_csv('/kaggle/input/datasets/pranavsinghdeveloper/desi-cs-student-survival/semester_records.csv')
```

```python
M=S.merge(D[['student_id','archetype','college_tier']],on='student_id')
M.groupby('semester').sgpa.agg(['mean','std','count']).round(2)
```

```python
fig,ax=plt.subplots(1,2,figsize=(13,4))
sns.lineplot(data=M,x='semester',y='sgpa',hue='archetype',ax=ax[0],errorbar=None); ax[0].set_title('SGPA by persona')
sns.lineplot(data=M,x='semester',y='backlogs_in_sem',ax=ax[1],color='crimson'); ax[1].set_title('Backlogs per student-semester'); plt.tight_layout(); plt.show()
```

## Estimate the slump per student (fixed-effects style)

```python
# within-student demeaning = fixed effects: how far is each semester from that student's own average?
M['dev']=M.sgpa-M.groupby('student_id').sgpa.transform('mean')
eff=M[M.groupby('student_id').semester.transform('max')==8].groupby('semester').dev.mean().round(3); print(eff)
eff.plot.bar(title='Deviation from own average (final-years, 8 semesters)',color='#2a9d8f'); plt.show()
```
