# 🍛 NLP on mess reviews and bunk excuses
Two tiny text columns: `mess_review_text` (sentiment ↔ `mess_rating`) and `reason_for_last_bunk` (↔ persona).

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns, warnings
warnings.filterwarnings('ignore'); sns.set_theme(style='whitegrid')
D = pd.read_csv('/kaggle/input/datasets/pranavsinghdeveloper/desi-cs-student-survival/students.csv')
S = pd.read_csv('/kaggle/input/datasets/pranavsinghdeveloper/desi-cs-student-survival/semester_records.csv')
T=D.dropna(subset=['mess_review_text']); print(T[['mess_rating','mess_review_text']].sample(5,random_state=3).to_string())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer; from sklearn.linear_model import Ridge, LogisticRegression
from sklearn.model_selection import train_test_split; from sklearn.pipeline import make_pipeline; from sklearn.metrics import r2_score, accuracy_score
a,b,c,d=train_test_split(T.mess_review_text,T.mess_rating,random_state=0)
p=make_pipeline(TfidfVectorizer(ngram_range=(1,2),min_df=2),Ridge()).fit(a,c); print('Rating R2:',round(r2_score(d,p.predict(b)),3))
v=p[0]; co=pd.Series(p[1].coef_,index=v.get_feature_names_out()); print('Most positive:',list(co.nlargest(8).index)); print('Most negative:',list(co.nsmallest(8).index))
```

## Bunk excuse → persona

```python
a,b,c,d=train_test_split(D.reason_for_last_bunk,D.archetype,random_state=0,stratify=D.archetype)
q=make_pipeline(TfidfVectorizer(ngram_range=(1,2)),LogisticRegression(max_iter=2000)).fit(a,c); print('Accuracy:',round(accuracy_score(d,q.predict(b)),3))
print(pd.crosstab(d,q.predict(b)).to_string())
```
